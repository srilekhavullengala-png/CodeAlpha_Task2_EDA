## Average Scores

Calculating the average scores in Mathematics, Reading, and Writing.

```python
df[['math score','reading score','writing score']].mean()
```

## Maximum Scores

```python
df[['math score','reading score','writing score']].max()
```

## Minimum Scores

```python
df[['math score','reading score','writing score']].min()
```

## Gender-wise Performance Analysis

Calculating the average scores of male and female students.

```python
df.groupby('gender')[['math score','reading score','writing score']].mean()
```

Female students perform better in Reading and Writing, while male students perform slightly better in Mathematics.

## Correlation Analysis

Calculating the correlation between Mathematics, Reading, and Writing scores.

```python
df[['math score','reading score','writing score']].corr()
```

Reading and Writing scores have a strong positive correlation, indicating that students who perform well in Reading also tend to perform well in Writing.
